```kotlin
// REQ-0001
// Provision to read the CSV fi le dpwh_fl ood_control_projects.csv containing 9,800+ rows of fl ood mitigation projects.

val df = DataFrame.readCSV("dpwh_flood_control_projects.csv")
df
```

```kotlin
// check distinct contractors

df.select("Contractor").distinct().print()
```

```kotlin
val df2 = df.add("BaseContractor") { row ->
    val name = row["Contractor"]?.toString()?.trim() ?: ""
    name.replace(Regex("\\(FORMERLY.*", RegexOption.IGNORE_CASE), "")
        .trim()
}


df2.select("Contractor", "BaseContractor").distinct().print(20)
```

```kotlin
val before = df["Contractor"].distinct().count()
val after = df2["BaseContractor"].distinct().count()
println("Contractors before: $before")
println("After removing (FORMERLY ...): $after")
println("Collapsed names: ${before - after}")
```

```kotlin
val mergedContractors = df2.groupBy("BaseContractor").aggregate {
    countDistinct("Contractor") into "VariantCount"
    countDistinct("ProjectId") into "ProjectCount"
}.filter { it["VariantCount"] as Int > 1 }

mergedContractors.print(15)
```

```kotlin
df2.filter { it["BaseContractor"] == "HI-TONE CONSTRUCTION & DEVELOPMENT CORP." }
    .select("Contractor", "ProjectId", "TypeOfWork", "FundingYear")
    .print(20)
```

```kotlin
val preMerge = df.groupBy("Contractor").aggregate {
    mean("CompletionDelayDays") into "AvgDelay"
    sum("CostSavings") into "TotalSavings"
}

val postMerge = df2.groupBy("BaseContractor").aggregate {
    mean("CompletionDelayDays") into "AvgDelay"
    sum("CostSavings") into "TotalSavings"
}

println("Pre-merge contractors: ${preMerge.rowsCount()}")
println("Post-merge contractors: ${postMerge.rowsCount()}")
```

```kotlin
mergedContractors.sortByDesc("VariantCount").print(10)
```

```kotlin
val overlapCheck = df2.groupBy("BaseContractor", "FundingYear").aggregate {
    countDistinct("Contractor") into "ContractorCount"
}.filter { (it["ContractorCount"] as Int) > 1 }

overlapCheck.sortByDesc("ContractorCount").print(10)
```
